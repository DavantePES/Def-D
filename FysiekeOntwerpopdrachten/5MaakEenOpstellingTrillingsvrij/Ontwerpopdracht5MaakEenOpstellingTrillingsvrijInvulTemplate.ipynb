## Opdracht 2b: Opsplitsen taak 2: afleiding ontwerpeis

```python
# Bereken hier je benodigde kantelfrequentie
g = 9.81 # zwaartekracht in m/s^2
m = 0 # massa in kg
c = 0 # veerconstante in N/m

f = 5 # Hz
omega = 2 * np.pi * f

# Afleiden formule -> 
omega0 = omega /2
print('omega0 = ', omega/2)

f0 = omega0 / (2 * np.pi)
print('Dus de kantelfrequentie is',f0, 'Hz')
```

```python
# Voeg hier je plot in.

x = np.linspace(0, 100, 1000) # frequentie in Hz
y = np.abs(1 / (1 - (x/(2*np.pi*f0))**2))

plt.plot(x/(2*np.pi), y)
plt.loglog()
plt.ylabel('Am/Av')
plt.xlabel('Frequentie (Hz)')
plt.xlim(0.00001, 10)
```

```python
# Bereken hier de minimale Delta-x waarbij je aan de eis van de opdracht voldoet.

deltax = g / omega0**2
print('deltax = ', deltax)
```

## Opdracht 3: Plot de grafiek.

```python
# Voeg hier je plot in. Lees goed de opdracht wat er in de grafiek moet staan!
# Het is ook handig om je grafiek op te slaan, doe dat door met plt.savefig().

m = np.linspace(0.1, 10, 100) # massa in kg
c = m*g/deltax # veerconstante in N/m
c2 = m*g/(deltax*2)
c5 = m*g/(deltax*5)

plt.figure()
plt.plot(m, c, label='deltax')
plt.plot(m, c2, label='deltax*2')
plt.plot(m, c5, label='deltax*5')
plt.xlabel('Massa (kg)')
plt.ylabel('Veerconstante (N/m)')
plt.xlim(0, 10)
plt.legend()
```

## Opdracht 7: Plotten maar

```python
# Voeg hier je plot in.

a_verhoudingen = np.array([1.60/0.93, 0.94/3.28, 1.26/5.88])
frequenties = np.array([1.0, 3.1, 5.1])

a_verhoudingen2 = np.array([0.64/0.14, 0.24/1.17, 0.37/3.28])

x = np.linspace(0, 100, 1000) # frequentie in Hz
y = np.abs(1 / (1 - (x/(2*np.pi*f0))**2))

y2 = np.abs(1 / (1 - (x/(2*np.pi*1.81))**2))

y3 = np.abs(1 / (1 - (x/(2*np.pi*1.25))**2))

plt.plot(x/(2*np.pi), y, label='Theoretisch Ontwerpgrafiek')
plt.plot(x/(2*np.pi),y2, 'r', label='Ontwerpgrafiek Data 1')
plt.plot(x/(2*np.pi), y3, 'g', label='Ontwerpgrafiek Data 2')
plt.plot(frequenties, a_verhoudingen, 'ro', label='Data 1')
plt.plot(frequenties, a_verhoudingen2, 'go', label='Data 2')
plt.plot(x, x*0+1/3, '--', label='Vereiste Demping')
plt.errorbar(frequenties, a_verhoudingen, yerr=0.1*a_verhoudingen, xerr=0.15*frequenties, fmt='ro')
plt.errorbar(frequenties, a_verhoudingen2, yerr=0.1*a_verhoudingen, xerr=0.15*frequenties, fmt='go')
plt.legend()
plt.loglog()
plt.ylabel('Am/Av')
plt.xlabel('Frequentie (Hz)')
plt.xlim(0.1, 15)
plt.ylim(0.1, 2000)
plt.show()
```
