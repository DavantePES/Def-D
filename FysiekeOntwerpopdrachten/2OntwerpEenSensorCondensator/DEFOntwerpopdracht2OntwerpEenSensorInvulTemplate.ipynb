```python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

epsilon0 = 8.854e-12  # permittiviteit van vacuum (F/m)

# Schrijf hier een python functie die de capaciteit van een plaatcondensator uitrekent. Zorg dat de functie
# als input de ontwerpparameters van de vlakke plaat formule accepteert (oppervlakte, afstand en diëlektrische
# constante) en als output de capaciteit van de vlakke plaat condensator met die ontwerpparameters.
# Je gaat deze formule de rest van de dag gebruiken: controleer hem goed!

def vlakkeplaat(A,d, epsilon):
    C = epsilon * A/d
    return C

# Optioneel: schrijf je functie zodanig dat je arrays als input kan geven en (dus) ook arrays als output.
```

```python
# Gebruik je formule om uit te rekenen wat je verwacht dat de capaciteit van jullie vlakke plaat condensatoren
# gaat zijn.

epsilon_polypropyleen = 2.2 * epsilon0
d = 0.0006
A = 0.1**2
C_berekend = vlakkeplaat(A, d, epsilon_polypropyleen)
print(C_berekend / 10**-12, "pF. Dit is de  berekende capaciteit van onze plaatcondensator.")

# C = [functieNaam(2.0), functieNaam(4.7),functieNaam(1.2)]


# print(f"Berekende capaciteit plaatcondensator groep XX is {C[0]:.1f} pF")
C_gemeten = 1.7/(1000*8*2*np.pi*100000)
print(C_gemeten / 10**-12, "pF. Dit is de gemeten capaciteit van onze plaatcondensator.")
```

## Opdracht 7: Plot de kalibratiegrafiek.

```python
# Plot hier je gemeten kalibratiegrafiek en plot in dezelfde figuur ook je verwachte ontwerpgrafiek.
C_meting = np.array([253.4, 225.1, 212.3, 199.7, 189.0, 157.1, 117.3, 95.1]) # in pF
M_meting = np.array([0, 0.111, 0.2744, 0.436, 0.4899, 0.642, 0.7693, 0.9725]) # in kg

# De reden dat dit een lineaire fit is is uitgelegd in opdracht 4.
def lineaire_fit(M, a, b):
    return a * M + b

popt, pcov = curve_fit(lineaire_fit, M_meting, C_meting)
a_fit, b_fit = popt

x_fit = np.linspace(0, 1, 100)
y_fit = lineaire_fit(x_fit, a_fit, b_fit)

plt.plot(x_fit, y_fit, label='lineaire fit')
plt.errorbar(M_meting, C_meting, yerr=10, xerr=0.002, fmt='ro', ecolor='red', alpha=0.5, label='Data')
plt.legend()
plt.title('Sensor kalibratiegrafiek')
plt.xlabel('$M$ (kg)')
plt.ylabel('$C$ (pF)')
plt.xlim(0, 1)
plt.ylim(0, 300)

uncertainty = np.sqrt(np.diag(pcov))/np.abs(a_fit)
print(f"De onzekerheden in a en b zijn respectievelijk {uncertainty[0]} en {uncertainty[1]}")
```

## Opdracht 8: Vul de tekst aan.

Onze sensor kan nog verbetered worden door de volgende verbeteringen (minimaal 2):

* Op de plek waar de bovenste plaat opzij schuift door de massa gaat dat best stroef. Hier kan een katrol toegevoegd worden zodat hij makkelijker schuift en dus een accurater capaciteit tegen gewicht geeft.
* De bovenste plaat schuift ook een beetje links/rechts boven de onderste, hierdoor verandert het oppervlak (en dus de capaciteit) ook. 
Een oplossing is een soort muurtje links en rechts van de plaat te maken zodat de plaat minder opzij schuift.
